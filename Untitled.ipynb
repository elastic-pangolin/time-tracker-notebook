```python
sync_day = "Apr 20"
# ==== SYNC TIME IN TICKETS TO JIRA WORKLOGS =====
import getpass

jiratoken = None
jiraserver = None
sync_df = df[df['day'] == sync_day]
tickets_per_day = ticket_summary(sync_df)

for day, ticket_agg in tickets_per_day:
# should only have one entry
    for key, value in ticket_agg.items():
        if key == "OTHER":
            # cannot log other activities right now
            print(f"please manually log {round(value/60,1)} hours 'other operational tasks' on {day}")
        else:
            value_rounded = int(round(value,0))
            conformation = False
            confirmation = (input(f"Send timelog of {value_rounded} minutes in {key} to jira? y/N").lower() == "y")
            if confirmation:
                if not jiraserver:
                    jiraserver = input("Please provide the jira web address ('https://...'): ")
                if not jiratoken:
                    jiratoken = getpass.getpass("Please provide a jira PAT: ")
                parsed_date = datetime.strptime(sync_day, "%b %d")
                sync_date = datetime.now().replace(day=parsed_date.day, month=parsed_date.month)
                jira_log_time(jiraserver, jiratoken, key, sync_date, value_rounded)
```
